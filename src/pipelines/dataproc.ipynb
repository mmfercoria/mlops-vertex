```python
from google.cloud import dataproc_v1 as dataproc
import google.auth
```

```python
# Autenticación automática usando la cuenta de servicio asociada a la instancia de Vertex Workbench
credentials, project_id = google.auth.default()
```

```python
# Nombre del clúster y región
cluster_name = "test-cluster-dataproc"
region = "us-east4"
zone = "us-east4-a" 
```

```python
# Configurar el cliente de Dataproc
cluster_client = dataproc.ClusterControllerClient(
    client_options={"api_endpoint": f"{region}-dataproc.googleapis.com:443"}
)
```

```python
# Configurar el clúster
cluster_config = {
    "project_id": project_id,
    "cluster_name": cluster_name,
    "config": {
        "gce_cluster_config": {
            "zone_uri": zone,
            "service_account": "815942598901-compute@developer.gserviceaccount.com",
        },
        "master_config": {
            "num_instances": 1,
            "machine_type_uri": "n1-standard-4",
        },
        "worker_config": {
            "num_instances": 2,
            "machine_type_uri": "n1-standard-4",
        },
        "software_config": {
            "image_version": "2.0-debian10",  # Imagen compatible con PySpark
            "optional_components": ["JUPYTER"],  # Incluir componentes como PySpark, Jupyter y Anaconda si los necesitas
            "properties": {
                "dataproc:pyspark.python": "/usr/bin/python3"  # Especifica el uso de Python 3 para PySpark
            }
        },
    },
}
```

```python
# Crear el clúster
operation = cluster_client.create_cluster(request={"project_id": project_id, "region": region, "cluster": cluster_config})
print("Creando el clúster con soporte para PySpark...")
```

```python
# Esperar a que la operación se complete
result = operation.result()
print(f"Clúster creado: {result.cluster_name}")
```
