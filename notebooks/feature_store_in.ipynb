```python
from google.cloud import aiplatform
from google.cloud import bigquery
```

```python
# Configuración de parámetros
project_id = "capable-hash-432501-a6"
location = "us-east4"
```

```python
# Define los parámetros de la Feature Store y el Feature Group (Entity Type)
featurestore_id = "featurestore_modern_iris"  # Nombre de tu Feature Store
entity_type_id = "plants_modern"  # Nombre del Feature Group (Entity Type)
entity_type_description = "Test feature group on modern vertex feature store"
```

```python
# Inicialización de AI Platform
aiplatform.init(
    project=project_id, 
    location=location
)
```

```python
# Crea o conecta la Feature Store
feature_store = aiplatform.Featurestore.create(
    featurestore_id=featurestore_id,  # Crea una nueva Feature Store o usa una existente
    #online_store_fixed_node_count=1  # Define nodos para almacenar características en tiempo real
)
```

```python
# Crea el Feature Group (Entity Type)
entity_type = feature_store.create_entity_type(
    entity_type_id=entity_type_id,
    description=entity_type_description,
)
```

```python
# Define las características (features) dentro del Feature Group basadas en el dataset de Iris
features = [
    {
        "feature_id": "sepal_length_cm",
        "value_type": "DOUBLE",
        "description": "Length of the sepal in cm"
    },
    {
        "feature_id": "sepal_width_cm",
        "value_type": "DOUBLE",
        "description": "Width of the sepal in cm"
    },
    {
        "feature_id": "petal_length_cm",
        "value_type": "DOUBLE",
        "description": "Length of the petal in cm"
    },
    {
        "feature_id": "petal_width_cm",
        "value_type": "DOUBLE",
        "description": "Width of the petal in cm"
    },
    {
        "feature_id": "target",
        "value_type": "INT64",
        "description": "Species of the iris flower"
    }
]
```

```python
# Crea las características dentro del Feature Group
for feature in features:
    entity_type.create_feature(
        feature_id=feature["feature_id"],
        value_type=feature["value_type"],
        description=feature["description"],
    )

print(f"Features added successfully to Feature Group '{entity_type_id}'.")
```
