```python
from google.cloud import aiplatform_v1beta1 as aiplatform
from google.cloud.aiplatform_v1beta1 import FeaturestoreServiceClient
from google.cloud.aiplatform_v1beta1.types import ImportFeatureValuesRequest, FeatureValueDestination, BigQuerySource
```

```python
# Inicializa Vertex AI
PROJECT_ID = "capable-hash-432501-a6"
LOCATION = "us-east4"  
FEATURESTORE_ID = "featurestore_modern_iris"
ENTITY_TYPE_ID = "plants_modern"
```

```python
aiplatform.FeaturestoreServiceClient()
```

```python
# Ruta del Featurestore y el EntityType (Feature Group)
featurestore_name = f"projects/{PROJECT_ID}/locations/{LOCATION}/featurestores/{FEATURESTORE_ID}"
entity_type_name = f"{featurestore_name}/entityTypes/{ENTITY_TYPE_ID}"
```

```python
# Inicializa el cliente de Featurestore
client = FeaturestoreServiceClient()
```

```python
# Configuración de la fuente de BigQuery
bigquery_source = BigQuerySource(
    input_uri="bq://capable-hash-432501-a6.iris_test.iris_data"  # Ruta de BigQuery a la tabla
)
```

```python
# Configuración del destino para las características
feature_time = None  # sin columna de tiempo

# Asigna las características y columnas de tu tabla
feature_specs = [
    ImportFeatureValuesRequest.FeatureSpec(id="sepal_length_cm", source_field="sepal_length_cm"),
    ImportFeatureValuesRequest.FeatureSpec(id="sepal_width_cm", source_field="sepal_width_cm"),
    ImportFeatureValuesRequest.FeatureSpec(id="petal_length_cm", source_field="petal_length_cm"),
    ImportFeatureValuesRequest.FeatureSpec(id="petal_width_cm", source_field="petal_width_cm"),
    ImportFeatureValuesRequest.FeatureSpec(id="target", source_field="target")  # Columna de la especie
]
```

```python
# Crear la solicitud de ingesta
request = ImportFeatureValuesRequest(
    entity_type=entity_type_name,  # Ruta del Feature Group
    bigquery_source=bigquery_source,  # La fuente de BigQuery
    feature_specs=feature_specs,  # Especificación de características
    entity_id_field="entity_id",  # Campo de ID único
    feature_time=feature_time  # Campo de tiempo, si lo tienes
)
```

```python
# Ejecuta la ingesta de datos
response = client.import_feature_values(request=request)
print(f"Import Feature Values Response: {response}")
```
